```python
from anthropic import Anthropic, APIError
from dotenv import load_dotenv

load_dotenv()

client = Anthropic()
model = "claude-sonnet-5"

def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

# temperature is gone on current-generation models (and dropped from the SDK
# entirely) — output_config.effort is the modern lever for controlling how
# much the model thinks before answering: low/medium/high/xhigh/max.
def chat(messages, system=None, effort="high"):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "output_config": {"effort": effort},
    }

    if system:
        params["system"] = system

    message = client.messages.create(**params)
    return message.content[0].text

messages = []

add_user_message(messages, "Generate an one sentence movie idea")

answer = chat(messages)

answer = chat(messages, effort="high")
```
